```python
"""
Dijkstra's Algorithm
"""

import heapq

class Graph:
    def __init__(self, vertices):
        self.V = vertices
        self.graph = [[] for _ in range(vertices)]

    def add_edge(self, u, v, w):
        self.graph[u].append((v, w))

    def dijkstra(self, src):
        # Init distances
        dist = {i: float('infinity') for i in range(self.V)}
        dist[src] = 0

        # Priority queue
        pq = [(0, src)]
        visited = set()

        while pq:
            # Get minimum distance vertex
            curr_dist, u = heapq.heappop(pq)

            # Skip if processed
            if u in visited:
                continue

            visited.add(u)

            # Update neighbors
            for v, weight in self.graph[u]:
                if v not in visited and dist[u] + weight < dist[v]:
                    dist[v] = dist[u] + weight
                    heapq.heappush(pq, (dist[v], v))

        return dist


def test_dijkstra():
    print("\n=== Testing Dijkstra's Algorithm ===")

    # Create test graph
    g = Graph(6)
    g.add_edge(0, 1, 5)
    g.add_edge(0, 2, 2)
    g.add_edge(1, 3, 3)
    g.add_edge(1, 4, 6)
    g.add_edge(2, 1, 1)
    g.add_edge(2, 3, 7)
    g.add_edge(2, 4, 4)
    g.add_edge(3, 4, 1)
    g.add_edge(3, 5, 5)
    g.add_edge(4, 5, 2)

    # Run algorithm
    distances = g.dijkstra(0)

    # Print results
    print("Dijkstra's Algorithm Results:")
    print("Shortest distances from source vertex 0:")
    for vertex, distance in distances.items():
        print(f"Vertex {vertex}: {distance}")

    # Validate results
    expected = {0: 0, 1: 3, 2: 2, 3: 6, 4: 6, 5: 8}
    print("\nValidation:")
    for vertex, distance in expected.items():
        result = "Correct" if abs(distances[vertex] - distance) < 1e-9 else f"Wrong (Expected: {distance})"
        print(f"Vertex {vertex}: {distances[vertex]} - {result}")


if __name__ == "__main__":
    test_dijkstra()
```

```python
"""
Bellman-Ford Algorithm
"""

class Graph:
    def __init__(self, vertices):
        self.V = vertices
        self.graph = [[] for _ in range(vertices)]

    def add_edge(self, u, v, w):
        self.graph[u].append((v, w))

    def bellman_ford(self, src):
        # Initialize distances
        dist = {i: float('infinity') for i in range(self.V)}
        dist[src] = 0

        # Relax edges repeatedly
        for _ in range(self.V - 1):
            for u in range(self.V):
                for v, w in self.graph[u]:
                    if dist[u] != float('infinity') and dist[u] + w < dist[v]:
                        dist[v] = dist[u] + w

        # Check for negative cycles
        has_negative_cycle = False
        for u in range(self.V):
            for v, w in self.graph[u]:
                if dist[u] != float('infinity') and dist[u] + w < dist[v]:
                    has_negative_cycle = True
                    break

        return dist, has_negative_cycle


def test_bellman_ford():
    print("\n=== Testing Bellman-Ford Algorithm ===")

    # Test 1: No negative cycles
    print("\nTest Case 1: Graph with no negative cycles")
    g1 = Graph(5)
    g1.add_edge(0, 1, 6)
    g1.add_edge(0, 3, 7)
    g1.add_edge(1, 2, 5)
    g1.add_edge(1, 3, 8)
    g1.add_edge(1, 4, -4)
    g1.add_edge(2, 1, -2)
    g1.add_edge(3, 2, -3)
    g1.add_edge(3, 4, 9)
    g1.add_edge(4, 0, 2)
    g1.add_edge(4, 2, 7)

    distances, has_negative_cycle = g1.bellman_ford(0)

    print("Bellman-Ford Algorithm Results:")
    print(f"Shortest distances from source vertex 0:")
    for vertex, distance in distances.items():
        print(f"Vertex {vertex}: {distance}")
    print(f"Negative cycle detected: {has_negative_cycle}")

    # Validation
    expected = {0: 0, 1: 6, 2: 2, 3: 7, 4: 2}
    print("\nValidation:")
    for vertex, distance in expected.items():
        result = "Correct" if abs(distances[vertex] - distance) < 1e-9 else f"Wrong (Expected: {distance})"
        print(f"Vertex {vertex}: {distances[vertex]} - {result}")

    # Test 2: With negative cycle
    print("\nTest Case 2: Graph with a negative cycle")
    g2 = Graph(4)
    g2.add_edge(0, 1, 1)
    g2.add_edge(1, 2, -1)
    g2.add_edge(2, 3, -1)
    g2.add_edge(3, 1, -1)  # Creates a negative cycle

    distances, has_negative_cycle = g2.bellman_ford(0)
    print(f"Negative cycle detected: {has_negative_cycle}")
    if has_negative_cycle:
        print("Distances are not reliable due to the presence of a negative cycle.")


if __name__ == "__main__":
    test_bellman_ford()
```

```python
"""
Floyd-Warshall Algorithm
"""

class Graph:
    def __init__(self, vertices):
        self.V = vertices
        self.graph = [[] for _ in range(vertices)]

    def add_edge(self, u, v, w):
        self.graph[u].append((v, w))

    def floyd_warshall(self):
        # Initialize distance matrix
        dist = [[float('infinity') for _ in range(self.V)] for _ in range(self.V)]

        # Distance to self is 0
        for i in range(self.V):
            dist[i][i] = 0

        # Set direct edges
        for u in range(self.V):
            for v, weight in self.graph[u]:
                dist[u][v] = weight

        # Main algorithm
        for k in range(self.V):
            for i in range(self.V):
                for j in range(self.V):
                    if dist[i][k] != float('infinity') and dist[k][j] != float('infinity'):
                        if dist[i][k] + dist[k][j] < dist[i][j]:
                            dist[i][j] = dist[i][k] + dist[k][j]

        return dist


def test_floyd_warshall():
    # Create test graph
    g = Graph(4)
    g.add_edge(0, 1, 3)
    g.add_edge(0, 3, 5)
    g.add_edge(1, 0, 2)
    g.add_edge(1, 2, 4)
    g.add_edge(1, 3, 8)  # This edge was incorrect in the previous version
    g.add_edge(2, 0, 1)
    g.add_edge(2, 3, 4)
    g.add_edge(3, 2, 2)

    # Run algorithm
    distances = g.floyd_warshall()

    # Print results
    print("\n=== Testing Floyd-Warshall Algorithm ===")
    print("Floyd-Warshall Algorithm Results:")
    for i in range(len(distances)):
        for j in range(len(distances[i])):
            print(f"From {i} to {j}: {distances[i][j]}")

    # Validate results
    expected = [
        [0, 3, 7, 5],
        [2, 0, 4, 8],
        [1, 4, 0, 4],
        [3, 6, 2, 0]
    ]

    print("\nValidation:")
    for i in range(len(expected)):
        for j in range(len(expected[i])):
            if abs(distances[i][j] - expected[i][j]) > 1e-9:
                print(f"Error at ({i}, {j}): Got {distances[i][j]}, Expected {expected[i][j]}")


if __name__ == "__main__":
    test_floyd_warshall()
```
